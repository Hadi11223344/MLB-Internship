```python

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_breast_cancer

# Display configuration
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 50)
pd.set_option("display.width", 120)

# Plot configuration
sns.set_theme(style="whitegrid", context="notebook")

plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["figure.dpi"] = 100
plt.rcParams["savefig.dpi"] = 300
plt.rcParams["savefig.bbox"] = "tight"
```

```python
dataset = load_breast_cancer(as_frame=True)

df = dataset.frame.copy()

X = dataset.data.copy()
y = dataset.target.copy()

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python

print("Dataset Shape:", df.shape)

print("\nData Types:")
print(df.dtypes.value_counts())

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe().T)
```

```python
print("Missing Values:")
display(df.isnull().sum().sort_values(ascending=False))

print("\nTotal Missing Values:")
print(df.isnull().sum().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
print(dataset.target_names)

print("\nTarget Counts:")
print(y.value_counts())

print("\nTarget Proportions:")
print(y.value_counts(normalize=True) * 100)
```

```python
target_counts = y.value_counts().sort_index()
target_labels = [
    dataset.target_names[index]
    for index in target_counts.index
]

plt.figure(figsize=(8,5))
sns.barplot(
    x=target_labels,
    y=target_counts.values
)

plt.title("Breast Cancer Target Distribution")
plt.xlabel("Diagnosis")
plt.ylabel("Number of Samples")

plt.tight_layout()
plt.show()
```

```python

selected_features = [
    "mean radius",
    "mean texture",
    "mean perimeter",
    "mean area",
    "mean smoothness",
    "mean compactness",
    "mean concavity",
    "mean symmetry"
]

df[selected_features].hist(
    bins=30,
    figsize=(16, 12)
)

plt.suptitle("Selected Feature Distributions", y=1.02)

plt.tight_layout()
plt.show()
```

```python

selected_features = [
    "mean radius",
    "mean texture",
    "mean perimeter",
    "mean area"
]

fig, axes = plt.subplots(
    2,
    2,
    figsize=(14, 10)
)

axes = axes.flatten()

for feature, ax in zip(selected_features, axes):

    sns.histplot(
        data=df,
        x=feature,
        hue="target",
        bins=30,
        kde=True,
        element="step",
        stat="density",
        common_norm=False,
        ax=ax
    )

    ax.set_title(feature)

plt.tight_layout()
plt.show()
```

```python

selected_features = [
    "mean radius",
    "mean texture",
    "mean perimeter",
    "mean area"
]

fig, axes = plt.subplots(
    2,
    2,
    figsize=(14, 10)
)

axes = axes.flatten()

for feature, ax in zip(selected_features, axes):

    sns.boxplot(
        data=df,
        x="target",
        y=feature,
        ax=ax
    )

    ax.set_title(feature)

plt.tight_layout()
plt.show()
```

```python

feature_corr = X.corr()

plt.figure(figsize=(18, 14))

sns.heatmap(
    feature_corr,
    cmap="coolwarm",
    center=0,
    annot=False
)

plt.title("Feature Correlation Heatmap")

plt.tight_layout()
plt.show()
```

```python

target_corr = df.corr()["target"].drop("target")

target_corr = target_corr.sort_values()

plt.figure(figsize=(10, 12))

target_corr.plot(kind="barh")

plt.title("Feature Correlation with Encoded Target")
plt.xlabel("Correlation")

plt.tight_layout()
plt.show()
```

```python
corr_matrix = X.corr().abs()

upper_triangle = corr_matrix.where(
    np.triu(
        np.ones(corr_matrix.shape),
        k=1
    ).astype(bool)
)

high_corr_pairs = (
    upper_triangle
    .stack()
    .sort_values(ascending=False)
)

display(high_corr_pairs.head(20))
```

## EDA Findings

### Dataset Structure
- 569 observations and 30 numerical predictor features.
- Binary classification problem.
- Target encoding: 0 = malignant, 1 = benign.

### Feature Distributions
- Several features exhibit right-skewed distributions.
- Some features have broad ranges and potentially extreme observations.

### Class Separation
- Mean radius, mean perimeter and mean area demonstrate noticeable distributional differences between classes.
- Malignant observations generally exhibit higher size-related measurements.

### Outliers
- Boxplots identify several observations outside conventional IQR-based whisker boundaries.
- Outliers should not automatically be removed without investigating their validity.

### Correlation
- Radius, perimeter and area are strongly correlated.
- Several mean and worst measurements have overlapping information.
- Strong feature-to-feature correlation may influence model interpretation.

### Target Relationships
- Several size- and concavity-related features show strong negative correlations with the encoded target.
- Since 0 represents malignant and 1 represents benign, negative correlations generally indicate association with the malignant class.
- Correlation does not establish causation or directly determine model feature importance.
